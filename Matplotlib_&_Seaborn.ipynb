### Styling lines and markers

The `plt.plot` function supports many arguments for styling lines and markers:

- `color` or `c`: set the color of the line ([supported colors](https://matplotlib.org/3.1.0/gallery/color/named_colors.html))
- `linestyle` or `ls`: choose between a solid or dashed line
- `linewidth` or `lw`: set the width of a line
- `markersize` or `ms`: set the size of markers
- `markeredgecolor` or `mec`: set the edge color for markers
- `markeredgewidth` or `mew`: set the edge width for markers
- `markerfacecolor` or `mfc`: set the fill color for markers
- `alpha`: opacity of the plot


Check out the documentation for `plt.plot` to learn more: [https://matplotlib.org/api/_as_gen/matplotlib.pyplot.plot.html#matplotlib.pyplot.plot](https://matplotlib.org/api/_as_gen/matplotlib.pyplot.plot.html#matplotlib.pyplot.plot)

The cycler in Matplotlib is a tool to define cycling patterns for plot properties, like colors, line styles, markers, etc., so that when you plot multiple lines or elements, they automatically follow a repeatable style sequence. Think of it as a “recipe for what comes next” in your plots

````python
# You can control almost every aspect of diagram using .rcParams
````

````python
from cycler import cycler
import numpy as np

x = np.linspace(0, 10, 100)
y = [np.sin(x + i) for i in range(4)]

# Set a custom color cycle
plt.rcParams["axes.prop_cycle"] = cycler(color=["red", "green", "blue", "orange"])
# for i, yi in enumerate(y):
# enumerate(y) gives you both the index and the value each time through the loop.
# So:
# first time → i = 0, yi = y[0]
# second time → i = 1, yi = y[1]
# etc.
for i, yi in enumerate(y):
    plt.plot(x, yi, label=f"line {i+1}")
plt.legend()
plt.show()
````

````python
a = np.linspace(0,5,100)
b = [np.sin(a + i) for i in range (6)]
plt.rcParams['axes.prop_cycle'] = cycler(color = ['red','green','blue','orange','black','pink'])
for i , bi in enumerate(b):
    plt.plot(a,bi,label= f"line {i+1}")
plt.legend();
plt.show();
````

x### Customizing Seaborn Figures

Since Seaborn uses Matplotlib's plotting functions internally, we can use functions like `plt.figure` and `plt.title` to modify the figure.

````python
# setting back style for data backgroud
sns.set_style('ticks');
# setting background color of scatterplot
plt.figure(figsize=(12,8),facecolor = 'lightblue')
plt.title('Sepal Dimension')
sns.scatterplot(
    data = flowers_df,
    x = 'sepal_length',
    y = 'sepal_width',
    hue = 'species',
    s = 100
)
````

````python
# Setting theme for scatterplot
plt.figure(figsize = (12,10),facecolor = 'lightblue');
# palette="pastel" in Seaborn just chooses the colors for 
# your points (or lines) in a soft, light, pastel style.
# the colors are soft and light, not bold or bright
# palette=["red","green","blue"].
sns.set_theme(style = 'darkgrid',palette="pastel");
plt.title('Sepal Dimension')
sns.scatterplot(
    data = flowers_df,
    x = 'sepal_length',
    y = 'sepal_width',
    hue = 'species',
    s = 100
);
````

## Histogram

A histogram represents the distribution of data by forming bins along the range of the data and then drawing bars to show the number of observations that fall in each bin.

As an example, let's visualize the how the values of sepal width in the flowers dataset are distributed. We can use the `plt.hist` function to create a histogram.

````python
flowers_df.sepal_width
````

````python
plt.figure(figsize = (12,9),facecolor = 'lightblue');
sns.set_theme(style = 'whitegrid');
plt.title('Distribution of Sepal Width');
plt.hist(flowers_df.sepal_width);
````

### Controlling the size and number of bins

We can control the number of bins, or the size of each bin using the `bins` argument.

````python
plt.figure(figsize = (8,6),facecolor = 'pink');
plt.title('Distribution of Sepal width');
sns.set_theme(style = 'ticks');
plt.hist(flowers_df.sepal_width,bins = 5);
````

Notice how the less the bin the higher the number of occurance
cause the value combine together in each making more occurance in each bin

````python
# Specifying the boundaries of each bin using Numpy function 
plt.hist(flowers_df.sepal_width,bins = np.arange(2,5,0.25));
````

````python
# Setting specific bins or unequal bin sizes
plt.hist(flowers_df.sepal_width,bins = [1,3,4,4.5]);
````

### Multiple Histograms

Similar to line charts, we can draw multiple histograms in a single chart. We can reduce the opacity of each histogram, so the the bars of one histogram don't hide the bars for others.

Let's draw separate histograms for each species of flowers.

````python
setosa_df = flowers_df[flowers_df.species == 'setosa']
versicolor_df = flowers_df[flowers_df.species == 'versicolor']
virginica_df = flowers_df[flowers_df.species == 'virginica']
````

````python
plt.hist(setosa_df.sepal_width,alpha = 0.4, bins = np.arange(2,5,0.25));
plt.hist(versicolor_df.sepal_width, alpha = 0.4, bins = np.arange(2,5,0.25));
# alpha=0.4 means the histogram bars are 40% opaque (or 60% transparent)
# This is super useful when you overlay multiple histograms, 
# like your setosa and versicolor ones, so you can see both 
# distributions at the same time without one completely blocking the otherWe can also stack multiple histograms on top of one another.
````

We can also stack multiple histograms on top of one another.

````python
plt.title('Distribution of Sepal Width');
plt.hist([setosa_df.sepal_width,versicolor_df.sepal_width,virginica_df.sepal_width,],
        bins = np.arange(2,5,0.25),
        stacked = True);
plt.legend(['Setosa','Versicolor','Virginica']);
````

````python
plt.title('Distribution of Sepal Width');
plt.hist([setosa_df.sepal_width,versicolor_df.sepal_width,virginica_df.sepal_width,],
        bins = np.arange(2,5,0.25),
        stacked = False);
plt.legend(['Setosa','Versicolor','Virginica']);
````

## Bar Chart
Bar charts are quite similar to line charts i.e. 
they show a sequence of values, however a bar is 
shown for each value, rather than points connected by lines. 
We can use the `plt.bar` function to draw a bar chart.

````python
years = range(2000, 2006)
apples = [0.35, 0.6, 0.9, 0.8, 0.65, 0.8]
oranges = [0.4, 0.8, 0.9, 0.7, 0.6, 0.8]
````

````python
plt.bar(years,oranges);
````

Like histograms, bars can also be stacked on top of one another. 
We use the `bottom` argument to `plt.bar` to achieve this.

````python
plt.bar(years,apples);
plt.bar(years,oranges,bottom = apples);
````
